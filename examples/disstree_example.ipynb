# disstree example: distance-informed tree on synthetic data

```python
import matplotlib.pyplot as plt
import numpy as np

from traminex.disstree import (
    disstree_feature_importance,
    disstree_fit,
    disstree_leaf_summary,
    disstree_plot,
    disstree_predict,
)

# Build a simple block distance matrix for two clusters
D = np.array(
    [
        [0, 1, 1, 5, 5, 5],
        [1, 0, 1, 5, 5, 5],
        [1, 1, 0, 5, 5, 5],
        [5, 5, 5, 0, 1, 1],
        [5, 5, 5, 1, 0, 1],
        [5, 5, 5, 1, 1, 0],
    ]
)
y = np.array([0, 0, 0, 1, 1, 1])

res = disstree_fit(D, y, n_components=2, max_depth=2, random_state=0)
ax = disstree_plot(res)
plt.show()

# Summaries
print(disstree_leaf_summary(res))
print(disstree_feature_importance(res))

# Predict for two new samples closer to each cluster
D_new = np.array([[0.5, 0.5, 0.5, 6, 6, 6], [6, 6, 6, 0.5, 0.5, 0.5]])
pred = disstree_predict(res, D_new)
print("Predictions for new samples:", pred)
```
