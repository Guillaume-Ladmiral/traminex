# HSPELL overlap and interval closure demo

```python
import matplotlib.pyplot as plt
import pandas as pd

import traminex as tx

h = pd.DataFrame(
    {
        "id": ["p1"],
        "begin1": [1],
        "end1": [3],
        "status1": ["A"],
        "begin2": [2],
        "end2": [4],
        "status2": ["B"],
    }
)

# later priority, inclusive intervals
X_later = tx.HSPELL_to_STS(
    h,
    id_col="id",
    begin=["begin1", "begin2"],
    end=["end1", "end2"],
    status=["status1", "status2"],
    overlap_priority="later",
    interval_closed="both",
)
fig1, ax1 = tx.sequence_index_plot(X_later)
ax1.set_title("Overlap priority: later, interval: both")
plt.show()

# earlier priority
X_earlier = tx.HSPELL_to_STS(
    h,
    id_col="id",
    begin=["begin1", "begin2"],
    end=["end1", "end2"],
    status=["status1", "status2"],
    overlap_priority="earlier",
)
fig2, ax2 = tx.sequence_index_plot(X_earlier)
ax2.set_title("Overlap priority: earlier")
plt.show()

# left-closed intervals (end excluded)
X_left = tx.HSPELL_to_STS(
    h,
    id_col="id",
    begin=["begin1", "begin2"],
    end=["end1", "end2"],
    status=["status1", "status2"],
    overlap_priority="later",
    interval_closed="left",
)
fig3, ax3 = tx.sequence_index_plot(X_left)
ax3.set_title("Interval closed: left (end excluded)")
plt.show()

# subject-specific tmin/tmax example
h2 = pd.DataFrame(
    {
        "id": ["p1", "p2"],
        "begin1": [1, 1],
        "end1": [4, 4],
        "status1": ["A", "B"],
        "tmin": [2, 1],
        "tmax": [3, 4],
    }
)
X_bounds = tx.HSPELL_to_STS(
    h2,
    id_col="id",
    begin=["begin1"],
    end=["end1"],
    status=["status1"],
    tmin_col="tmin",
    tmax_col="tmax",
    fillblanks="_",
)
fig4, ax4 = tx.sequence_index_plot(X_bounds)
ax4.set_title("Subject-specific bounds via tmin_col/tmax_col")
plt.show()
```
